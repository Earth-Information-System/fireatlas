# FIRMS backfill 


This notebook shows how to re-download VIIRS active fire detections from FIRMS if we ever need to reset the data in FEDSinput. Rather than requesting one day at a time via the API, which is very inefficient and unlikely to work in one shot, we do one bulk download using the FIRMS bulk download tool, then split that into daily files ourselves. 

* 1: download the required bulk FIRMS download. [https://firms.modaps.eosdis.nasa.gov/download/](https://firms.modaps.eosdis.nasa.gov/download/) -> log in with Earthdata Login -> Create New Request -> World (important- don't select CONUS, as we expect all detections globally to be present in raw input files) -> your desired sensor source (FEDS currently supports only VIIRS) and date range -> .csv -> send email confirmation: yes -> Submit. 

* 2: when this download is ready, you will recieve an email to the account associated with your Earthdata login. This may take an hour or two. Download the resulting zip file to your local machine. 

* 3: Upload the zip file to your home directory (~) on MAAP Hub, then unzip it as shown below. Swap in the name of your zip file here.
